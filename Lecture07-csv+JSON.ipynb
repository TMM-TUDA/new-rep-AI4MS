```python
# IMPORTANT: files are read in as strings 
# 1st method: read in the csv file as a normal file 
f1 = open("hw25000.csv")
print(f1.readlines()[-95])
f1.close()
```

```python
# 1st method + data: get the data seperated into lists 
f1 = open("hw25000.csv")
f1.readline() # skip the header 
for xx in f1:
    dataset = xx.split(",")  # each line separated into a list, with three elements corresponding to 3 columns
    #print(dataset)
    #print(dataset[0].strip())
    print(len(dataset[1]),len(dataset[1].strip())) # use strip() to get rid of extra spaces before/after the numbers 
f1.close()
```

```python
# 1st method + data + conversion: get the data seperated into lists 
f1 = open("hw25000.csv")
f1.readline() # skip the header 

index=[]
height=[]
weight=[]

for line in f1:
    dataset = line.split(",")  # how if the data contains ,
    index.append(float(dataset[0].strip())) # 1st-> strip(), 2nd-> convert str to float, 3nd-> append the numbers to the list
    height.append(float(dataset[1].strip()))
    weight.append(float(dataset[2].strip()))   
f1.close()


from matplotlib import pyplot as plt
plt.hist(height,bins=100)
plt.hist(weight,bins=200)
plt.show() #needed for the display 
```

```python
# 2nd method: the second method by using the the csv module 
import csv
with open('hw25000.csv', 'r') as file:
    reader = csv.reader(file)
    type(reader)
    header = next(reader) #skip the header if needed
    data = [row for row in reader]
    print(header)
    print(data[99])
```

```python
# the second method by using the the csv module 
import csv

index=[]
height=[]
weight=[]
with open('hw25000.csv', 'r') as file:
    reader = csv.reader(file)
    header = next(reader) #skip the header if needed

    for row in reader:
        #print(row[1])
        index.append(float(row[0].strip()))
        height.append(float(row[1].strip()))
        weight.append(float(row[2].strip()))
    
    #print(weight)

#print(height)
#print(weight)
from matplotlib import pyplot as plt
plt.plot(weight, height)
plt.show()
```

```python
# 2nd method: to write to .csv files
file_2write = open("output3.csv",'w')
writer = csv.writer(file_2write)
writer.writerow(["index","height","weight"])
data = []
for i in range(len(index)-1):
    data.append([index[i],height[i],weight[i]])
    #writer.writerow(index[i],height[i],weight[i])

for i in range(len(index)-1):
    writer.writerow(data[i])
    
file_2write.close()
```

# .json files
Java script object notation = python dictionaries

```python
import json

# a JSON object is the same as a python dictionary except for: 
# true/false (Java) <-> True/False (Python)
# null (Java)       <-> None (Python)

test='''
{"title":"Gattaca",
     "release_year":1997,
     "is_awesome":true,
     "won_oscar":false,
     "actors":["Ethan Hawke", "Uma Thurman","Loren Dean"],
     "budget": null,
     "credits":{"director":"Andrew Niccol",
               "writer":"Andrew Niccol",
               "composoer":"Michael Nyman"}
     }
'''
movie = json.loads(test) # load a string into dictionary
#type(movie)
#movie["title"] #
movie["credits"]["director"]
```

```python
file = open("movie.json",'w')
json.dump(movie,file) # json.dumps is also possible 
file.close()
```
